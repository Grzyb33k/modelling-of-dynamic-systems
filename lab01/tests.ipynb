```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
```

```python
def f(x, y):
    return y ** 2

y0_arr = [-2]

sol = solve_ivp(f, [0, 0.48], y0_arr, method="RK45", rtol=1e-6, atol=1e-3, max_step=0.001)


for i, y0 in enumerate(y0_arr):
    plt.plot(sol.t, sol.y[i], label=y0)

plt.xlim(0, 1)
plt.ylim(-5, 10)

plt.legend()
```
