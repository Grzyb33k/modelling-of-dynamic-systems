# Wyznaczanie punktów stałych równania

## Równanie $\ddot{\theta} + \theta = 0$

Mamy układ równań

\begin{align}
    \dot{\theta} &= \omega \\
    \dot{\omega} &= -\theta
\end{align}

Dla którego występuje jeden punkt stały $(\theta,\omega) = (0, 0)$


## Równanie $\ddot{\theta} + \theta - \frac{1}{6} \theta^3 = 0$

Podstawiając za $\omega = \dot{theta}$ otrzymujemy

\begin{align}
    \dot{\theta} &= \omega \\
    \dot{\omega} &= \frac{1}{6}\theta^3 - \theta
\end{align}

Współrzędna $\omega$ punktów stałych jest równa $\omega=0$, a współrzędna $\theta$ spełnia równanie $\frac{1}{6}\theta^3 - \theta = 0$.

Zakładając, że $\theta = 0$ otrzymujemy punkt $(\theta,\omega) = (0, 0)$. Dla $\theta \neq 0$ mamy
$$
    \frac{1}{6}\theta^2 = 1,
$$
zatem $\theta = \pm \sqrt{6}$.

Mamy zatem kolejne dwa punkty stałe

\begin{align}
    (\theta,\omega) &= (\sqrt{6}, 0) \\
    (\theta,\omega) &= (-\sqrt{6}, 0)
\end{align}

## Równanie $\ddot{\theta} + \sin(\theta) = 0$

Układ równań po podstawieniu $\omega = \dot{\theta}$ sprowadza się do

\begin{align}
    \dot{\theta} &= \omega \\
    \dot{\omega} &= -\sin(\theta)
\end{align}

Ponownie, dla współrzędnej $\omega$ mamy $\omega=0$, natomiast dla $\theta$ mamy równanie $\sin(\theta) = 0$.
Sinus kąta $\theta$ jest równy zero dla wielokrotności $\pi$, zatem $\theta = k\pi$, dla $k \in \mathbb{Z}$.
Zbiorem punktów stałych jest zatem zbiór $(\theta, \omega) = (k\pi, 0)$ dla $k \in \mathbb{Z}$.

# Określanie stabilności punktów stałych - macierz Jacobiego

Stabilność punktów określa się badając wartości własne równania $det(J - \lambda) = 0$, gdzie
$$
 J = 
    \begin{bmatrix}
        \frac{\partial f(x,y)}{\partial x} & \frac{\partial f(x,y)}{\partial y} \\ 
        \frac{\partial g(x,y)}{\partial x} & \frac{\partial g(x,y)}{\partial y}
    \end{bmatrix}
$$
a $f(x,y)$ oraz $g(x,y)$ są funkcjami pochodnych z układu równań. 

## Równanie $\ddot{\theta} + \theta = 0$

Mamy $f(\theta,\omega) = \omega$ oraz $g(\theta,\omega) = -\theta$

Macierz Jacobiego

$$
    J = 
    \begin{bmatrix}
        0 & 1 \\
        -1 & 0
    \end{bmatrix}
$$

zatem problem własny sprowadza się do $\lambda^2 + 1 = 0 \Rightarrow \lambda = \pm i$ co oznacza, iż nie da się określić jednoznacznie stabilności tego punktu.

## Równanie $\ddot{\theta} + \theta - \frac{1}{6} \theta^3 = 0$

Mamy $f(\theta,\omega) = \omega$ oraz $g(\theta,\omega) = \frac{1}{6} \theta^3 - \theta$.

Macierz Jacobiego

$$
    J = 
    \begin{bmatrix}
        0 & 1 \\
        \frac{1}{2} \theta^2 - 1 & 0
    \end{bmatrix}
$$

wtedy $\lambda^2 - \frac{1}{2} \theta^2 + 1 = 0$

Dla punktu stałego $(0, 0)$ mamy $\lambda = \pm i$, natomiast dla pary punktów $(\pm\sqrt{6}, 0)$ otrzymujemy $\lambda^2 = 2 \Rightarrow \lambda = \pm \sqrt{6}$. Mamy zatem dla $(\pm\sqrt{6}, 0)$ sytuację, że $\lambda_1\lambda_2 < 0$ więc jest to tzw. saddle point.

## Równanie $\ddot{\theta} + \sin(\theta) = 0$

Mamy $f(\theta,\omega) = \omega$ oraz $g(\theta,\omega) = -\sin(\theta)$.

Macierz Jacobiego

$$
    J = 
    \begin{bmatrix}
        0 & 1 \\
        -\cos(\theta) & 0
    \end{bmatrix}
$$

Równanie dla wartości własnych $\lambda^2 + \cos(\theta) = 0 \Rightarrow \lambda = \pm i\sqrt{cos(\theta)}$ co dla punktów stałych $\theta=k\pi$ sprawia, że $\cos(\theta) = \pm 1$. Mamy zatem dwa przypadki.

### $\theta = 2n\pi, n \in \mathbb{Z}$

Wtedy $\cos(\theta) = 1$, zatem $\lambda = \pm i$ i ponownie nie można określić stabilności.

### $\theta = (2n+1)\pi, n \in \mathbb{Z}$

Wtedy $\cos(\theta) = -1$, zatem $\lambda = \pm 1$ i punkt ten jest tzw. saddle point.

# Portet fazowy

```python
thetas = np.linspace(-10, 10, 1000)
omegas = np.linspace(-5, 5, 1000)

THETA, OMEGA = np.meshgrid(thetas, omegas)

THETA_DOT_ARR = [OMEGA, OMEGA, OMEGA] 
OMEGA_DOT_ARR = [-THETA, 1/6 * THETA**3 - THETA, -np.sin(THETA)]

titles = [
    "$\\ddot{\\theta} + \\theta = 0$",
    "$\\ddot{\\theta} + \\theta - \\frac{1}{6} \\theta^3 = 0$",
    "$\\ddot{\\theta} + \\sin(\\theta) = 0$"
    
]

def make_streamplot(X, Y, X_DOT, Y_DOT):
    fig, ax = plt.subplots(dpi=200)

    ax.streamplot(X, Y, X_DOT, Y_DOT, density=2, linewidth=0.5, color='teal')

    ax.contour(X, Y, X_DOT, levels=[0], lw=0.5, colors='tomato')
    ax.contour(X, Y, Y_DOT, levels=[0], lw=0.5, colors='royalblue')

    ax.set_xlabel("$\\theta$")
    ax.set_ylabel("$\\omega$")

    ax.set_title(f"Rozwiązanie dla równania {titles[i]}")

    plt.gca().set_aspect('equal')

    plt.show()

for i in range(len(THETA_DOT_ARR)):
    make_streamplot(THETA, OMEGA, THETA_DOT_ARR[i], OMEGA_DOT_ARR[i])
```

Analizując wykresy widać, że punkty $(0,0)$ oraz punkty dla $\theta=(2n+1)\pi$ są punktami stabilnymi. Ponadto wszystkie wyznaczone punkty stabilne pokrywają się z przecięciami się izoklin. Dodatkowo widać wyraźnie różnice w portrecie fazowym dla przybliżenia małych kątów, gdzie uzyskuje się oscylacje harmoniczne. 

# Numeryczne wyznaczanie rozwiązania

```python
t_min, t_max = 0, 10
t_span = [t_min, t_max]
t = np.linspace(t_min, t_max, 100)

theta_0 = 0
omega_0_arr = np.arange(0.5, 3.01, 0.5)

colors = plt.cm.viridis(np.linspace(0, 1, len(omega_0_arr)))

def f1(t, y):
    theta, omega = y

    d_theta = omega
    d_omega = -theta

    return [d_theta, d_omega]

def f2(t, y):
    theta, omega = y
    
    d_theta = omega
    d_omega = 1/6 * theta ** 3 - theta

    return [d_theta, d_omega]

def f3(t, y):
    theta, omega = y
    
    d_theta = omega
    d_omega = -np.sin(theta)

    return [d_theta, d_omega]

f_arr = [f1, f2, f3]

solutions = np.zeros((len(f_arr), len(omega_0_arr)), dtype=np.ndarray)

for i, f in enumerate(f_arr):

    fig, (ax1, ax2) = plt.subplots(2, 1, dpi=200, sharex=True, constrained_layout=True)


    for j, omega_0 in enumerate(omega_0_arr):
        sol = solve_ivp(
            f,
            t_span,
            [theta_0, omega_0],
            method="DOP853",
            rtol=1e-6,
            atol=1e-3,
            max_step=0.001
        )

        solutions[i, j] = [sol.t, sol.y[0], sol.y[1]]

        ax1.plot(sol.t, sol.y[0], lw=1.0, label=f"$\\omega_0 =$ {omega_0:.1f}", color=colors[j])

        ax1.set_ylabel("$\\theta (t)$")
        ax1.legend(fontsize=8)
        ax1.set_title(f"$\\theta (t)$ dla {titles[i]}")

        ax2.plot(sol.t, sol.y[1], lw=1.0, label=f"$\\omega_0 =$ {omega_0:.1f}", color=colors[j])

        ax2.set_xlabel("$t$")
        ax2.set_xlim(t_min, t_max)
        ax2.set_ylabel("$\\omega (t)$")
        ax2.legend(fontsize=8)
        ax2.set_title(f"$\\omega (t)$ dla {titles[i]}")

        if i == 1:
            ax1.set_ylim(-3, 5)
            ax2.set_ylim(-3, 5)
```

Dla większych prędkości początkowych funkcja przybliżająca sinusa nie jest zbieżna, co widać po wykresie prędkości kątowej jak i kąta $\theta$. Dla przybliżenia liniowego otrzymujemy oscylacje harmoniczne, a dla rozwiązania pełnego rzeczywiste trajektorie wahadła matematycznego. Do tych samych wniosków prowadzi analiza trajektorii $\omega(\theta)$ na wykresach poniżej.

```python
for i in range(len(f_arr)):
    fig, ax = plt.subplots(dpi=200)

    for j, omega_0 in enumerate(omega_0_arr):

        res = solutions[i, j]

        theta, omega = res[1:]

        ax.plot(theta, omega, color=colors[j], lw=1.0, label=f"$\\omega_0 = $ {omega_0:.1f}")

    plt.gca().set_aspect('equal')

    ax.set_xlabel("$\\theta$")
    ax.set_ylabel("$\\omega$")

    if i == 1:
        ax.set_xlim(-5, 5)
        ax.set_ylim(-5, 5)

    ax.legend(fontsize=5)

    ax.set_title(f"Trajektorie dla równania {titles[i]}")
```

# Energia mechaniczna

```python
mass = 1.0
length = 9.81
g = 9.81

def total_energy(theta, omega, m, l, g):
    return m * g * l * (1 - np.cos(theta)) + 0.5 * m * l ** 2 * omega ** 2

for i in range(len(f_arr)):
    fig, (ax1, ax2) = plt.subplots(
        2, 1,
        figsize=(8, 6),
        dpi=200,
        constrained_layout=True
    )

    for j, omega_0 in enumerate(omega_0_arr):
        t, theta, omega = solutions[i, j]
        E = total_energy(theta, omega, mass, length, g)

        ax1.plot(
            t, E,
            color=colors[j],
            lw=1.0,
            label=rf"$\omega_0={omega_0:.1f}$"
        )

        if j == 0:
            dE = E - E[0]
            ax2.plot(t, dE, color=colors[j], lw=1.0)

    ax1.set_xlabel("$t$")
    ax1.set_ylabel("$E(t)$")
    ax1.set_title(f"Energia całkowita {titles[i]}")
    ax1.legend(fontsize=6)

    ax2.set_xlabel("$t$")
    ax2.set_ylabel(r"$\Delta E(t)=E(t)-E(0)$")
    ax2.set_title(r"Odchylenie energii dla $\omega_0=0.5$")

    plt.show()
```

Dla prędkości kątowej $\omega_0=0.5$ energia całkowita jest zachowana, a jej różnice między wartością początkową są znikome (rzędu $10^{-16}$).
