# Zadanie 1

Dane jest równanie
$$
    \dot{x} = x (1 - x^2)
$$

## Wyznaczenie punktów stałych równania

Punkty stałe są rozwiązaniem równania $f(x) = 0$, gdzie $f(x) = x(1 - x^2)$.

Rozwiązaniem są trzy punkty
\begin{itemize}
    \item $x_0 = 0$
    \item $x_1 = 1$
    \item $x_2 = -1$
\end{itemize}

Stabilność określa się badając znak pochodnej funkcji $f(x)$
$$
    f'(x) = 1 - 3x^2
$$
otrzymując odpowiednio
\begin{itemize}
    \item Dla $x_0 = 0$ mamy $f'(x_0) = 1 > 0$ - punkt niestabilny
    \item Dla $x_1 = 1$ mamy $f'(x_1) = -2 < 0$ - punkt stabilny
    \item Dla $x_2 = -1$ mamy $f'(x_2) = -2 < 0$ - punkt stabilny
\end{itemize}

## Linearyzacja i rozwiązania analityczne

Linearyzacja w otoczeniach punktów stałych wyznacza się z
$$
    \dot{x}(t) \approx f(x^*) + f'(x^*) \cdot (x - x^*)
$$

Dla $x^*=0$ mamy $\dot{x}(t) = x$, $x^*=1 \Rightarrow \dot{x}(t) = -2x+2$ oraz dla $x^*=-1 \Rightarrow \dot{x}(t) = -2x-2$.

Rozwiązanie analityczne otrzymuje się rozwiązując $\frac{dx}{dt} = \dot{x}(t)$ dla otrzymanych wyżej wyrażeń oraz uwzględniając warunek początkowy $x(t = 0) = x_0$.

Otrzymano następujące rozwiązania analityczne

\begin{itemize}
    \item Dla $x^*=0$ mamy $x(t) = x_0 e^t$
    \item Dla $x^*=1$ mamy $x(t) = 1 + (x_0 - 1)e^{-2t}$
    \item Dla $x^*=-1$ mamy $x(t) = (x_0 + 1)e^{-2t} - 1$
\end{itemize}

## Numeryczne rozwiązanie równania

Rozwiązanie równania $\dot{x} = x(1 - x^2)$ dla $t\in[0;3]$ dla przyjętych warunków początkowych $x(t=0) = x_0 = -2, -1.8, ... 1.8, 2.0$

```python
def f(t, x):
    return x * (1 - x**2)

t_min, t_max = 0, 3

t_span = [t_min, t_max]
t_eval = np.linspace(t_min, t_max, 100)

x0_vals = np.arange(-2.0, 2.01, 0.2)

fig, ax = plt.subplots(dpi=200)

for x0 in x0_vals:
    sol = solve_ivp(
        f,
        t_span,
        [x0],
        method="DOP853",
        rtol=1e-6,
        atol=1e-3,
        max_step=0.001)

    ax.plot(sol.t, sol.y[0], lw=1.0)

ax.set_xlim(t_min, t_max)
ax.set_xlabel("$t$")
ax.set_ylabel("$x(t)$")
plt.show()
```

## Porównanie rozwiązań numerycznych z analitycznym

```python
def f_lin(t, x0):
    if abs(x0 - 1) < 0.25:
        return 1 + (x0 - 1) * np.exp(-2*t)
    elif abs(x0) < 0.25:
        return x0 * np.exp(t)
    elif abs(x0 +1) < 0.25:
        return (x0 + 1) * np.exp(-2*t) - 1

x0_vals = [-1.1, -0.9, -0.1, 0.1, 0.9, 1.1]

t_min, t_max = 0, 3

t_span = [t_min, t_max]
t_eval = np.linspace(t_min, t_max, 100)


fig, ax = plt.subplots(dpi=200)

for x0 in x0_vals:
    sol = solve_ivp(
            f,
            t_span,
            [x0],
            method="DOP853",
            rtol=1e-6,
            atol=1e-3,
            max_step=0.001)

    y_lin = f_lin(t_eval, x0)

    ax.plot(sol.t, sol.y[0], lw=1.0, color='black')
    ax.plot(
        t_eval,
        y_lin,
        label="$\\tilde{x}_i(t)$ " f"dla $x_0 = $ {x0:.1f}",
        ls='--',
        lw=1.5
        )

ax.set_xlim(t_min, t_max)

ax.set_xlabel("$t$")
ax.set_ylabel("$x(t)$")

plt.legend(fontsize=8)

plt.show()
```

Na wykresie linią przerywaną zaznaczono rozwiązanie analityczne $\tilde{x_i}(t)$ dla zadanego parametru początkowego $x_0$, linia czarna ciągła odpowiada rozwiązaniu numerycznemu.

## Potencjał $V(x)$

```python
def V_x(x):
    return 0.25 * x**4 - 0.5 * x**2

x_stable_vals = [-1, 0, 1]

fig, ax = plt.subplots(dpi=200)

x = np.linspace(-2, 2, 100)

ax.plot(x, V_x(x), color='teal', lw=1.0)

for x_stable in x_stable_vals:
    ax.scatter(x_stable, V_x(x_stable), color='crimson', s=15, zorder=4)

ax.set_xlabel("$x$")
ax.set_ylabel("$V(x)$")
```

Na powyższym wykresie wyznaczono potencjał $V(x)$ dla zakresu w których znajdują się punkty stałe. Widać, że dla $x=-1$ oraz $x=1$ otrzymujemy punkty stałe stabilne, co zgadza się z analizą w poprzednim podpunkcie. Dla $x=0$ otrzymujemy punkt niestabilny.

# Zadanie 2

Dany jest układ równań

\begin{align}
    \dot{x} &= x(y-2) \\
    \dot{y} &= -3x + 3y + x^2 - y^2
\end{align}

## Wyznaczenie punktów stałych

Przyrównując pierwsze równanie do zera otrzymujemy $x(y-2)=0$, zatem $x=0 \lor y=2$

Dla $x=0$ z drugiego równania otrzymujemy $y(3-y)=0$, zatem $y=0 \lor y=3$. Rozwiązaniem jest para punktów $(x,y)= (0,0)$ oraz $(x,y)= (0,3)$.

Rozwiązanie drugiego równania dla $y=2$ daje kolejną parę punktów $(x,y) = (1,2)$ oraz $(x,y)= (2,2)$. 

## Portert fazowy

```python
xy_min, xy_max = -1, 4

stable_points = [(0, 0), (0, 3), (1, 2), (2, 2)]

x = np.linspace(xy_min, xy_max, 200)
y = np.linspace(xy_min, xy_max, 200)

X, Y = np.meshgrid(x, y)

X_DOT = X * (Y - 2)
Y_DOT = -3 * X + 3 * Y + X * X - Y * Y

fig, ax = plt.subplots(dpi=200)

ax.streamplot(X, Y, X_DOT, Y_DOT, density=2, linewidth=0.5, color='teal')

ax.contour(X, Y, X_DOT, levels=[0], lw=0.5, colors='tomato')
ax.contour(X, Y, Y_DOT, levels=[0], lw=0.5, colors='royalblue')

for point in stable_points:
    plt.scatter(point[0], point[1], s=15, color='crimson', zorder=4)
    plt.text(point[0], point[1]+0.1, f"({point[0]},{point[1]})", fontsize=8, ha='center', zorder=4)

ax.set_xlabel("$x$")
ax.set_ylabel("$y$")

plt.gca().set_aspect('equal')

plt.show()
```

Na wykresie zaznaczono zarówno punkty stałe wyznaczone wcześniej jak i izokliny (niebieska dla $\dot{x}=0$ oraz czerwona dla $\dot{y}=0)$. Widać, że wszystkie punkty stałe znajdują się w miejscach przecięcia się izoklin. Nie są to jednak punkty stałe stabilne. Analiza portretu fazowego wskazuje na to, iż jedynie punkt $(x,y) = (1,2)$ jest punktem stabilnym - wszystkie wektory skierowane są do punktu. W pozostałych przypadkach wektory pola pozwalają na "ucieczkę" z punktu stałego, nie są to zatem punkty stabilne.
